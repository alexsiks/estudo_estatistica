📈 Exemplo — Gráfico de Distribuição Normal
1. Criar os dados

Vamos utilizar um conjunto de valores:

$$ 10,\ 12,\ 14,\ 16,\ 18 $$

Esses valores serão utilizados para encontrar os parâmetros da distribuição.

```python
# Importando bibliotecas
import pandas as pd
import numpy as np
```

```python
# Gerando amostras.
df=pd.DataFrame({
    "amostra":[10,12,14,16,18]
    })

df
```

2. Calcular a média

Primeiro calculamos a média:

$$ \mu = \frac{10+12+14+16+18}{5} $$ $$ \boxed{\mu=14} $$

A média será o centro da distribuição normal.

```python
total_amostra=np.sum(df["amostra"])
quantidade_amostra=np.count_nonzero(df["amostra"])
media_amostra=total_amostra / quantidade_amostra

media_amostra
```

3. Calcular o desvio padrão

Para esses dados:

$$ \sigma \approx 2,83 $$

Portanto, temos:

Média: 14
Desvio padrão: 2,83

```python
# Caucule a diferença de cada elemento pela média.
diferenca_media=df["amostra"] - media_amostra
# Eleve a diferença ao quadrado.
diferenca_media_quadrado=diferenca_media ** 2
# Some a diferenca media ao quadrado.
soma_diferenca_media_quadrada=np.sum(diferenca_media_quadrado)
# caucule a variancia.
variancia = soma_diferenca_media_quadrada / quantidade_amostra

# Caucule o desvio padrão pela a raiz quadrada da variancia.

#------------------------------------------------------
# O desvio padrão representa o quanto as amostras se afastão da média.
#------------------------------------------------------
desvio_padrao = np.sqrt(variancia)
desvio_padrao
```
